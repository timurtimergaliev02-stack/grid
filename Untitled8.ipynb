```python
import openai
from openai import OpenAI
import os
from getpass import getpass
```

```python
os.environ['OPENAI_API_KEY'] = getpass(prompt='Введи свой API ключ')
```

```python
client = OpenAI(base_url='https://api.vsellm.ru/')
```

```python
model = client.chat.completions.create(model='gpt-4o-mini', messages=[{
    'role': 'system', 
    'content': 'Тебе нужно классифицировать входящий текст на 3 категории: отрицательный, нейтральный, положительный.'},{
    'role': 'user',
    'content': 'Я ненавижу машинное обучение'
    }])
```

```python
print(model.choices[0].message.content)
```

```python
model1 = client.chat.completions.create(model='gpt-4o-mini', messages=[{
    'role': 'system', 
    'content': 'Тебе нужно классифицировать входящий текст на 3 категории: отрицательный, нейтральный, положительный. Добавь также обьяснение твоего ответа'},
                                                                       {
    'role': 'user',
    'content': 'Я обожаю машинное обучение'
    }], temperature=0.5, stream=True)
```

```python
for i in model1:
    if i.choices:
        print(i.choices[0].delta.content, end="")
```

```python
from langchain_openai import ChatOpenAI
```

```python
model = ChatOpenAI(
    model='gpt-4o-mini', 
    base_url='https://api.vsellm.ru/', 
    model_kwargs={'seed': 365}, 
    temperature=0.1
)
```

```python
response = model.invoke('Как думаешь насколько тяжело сейчас найти работу в России')
```

```python
print(response.content)
```

## Разновидности Ролей и Контекста

```python
from langchain_core.messages import AIMessage, HumanMessage, SystemMessage
```

```python
message_H = HumanMessage(content='РАсскажи про обучение с подкреплением')
message_S = SystemMessage(content='Сделай обьяснение лаконичным с примерами и в менее строгой форме')
```

```python
response = model.invoke([message_H])
```

```python
print(response.content)
```

```python
response = model.invoke([message_H, message_S])
```

```python
print(response.content)
```
