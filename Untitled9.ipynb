```python
from catboost import CatBoostClassifier, Pool, CatBoostRegressor
from sklearn.model_selection import train_test_split
import numpy as np
```

```python
X = df.drop('target', axis=1)
y = df['target'].fillna(0)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
model = CatBoostClassifier(
    random_state=42,
    thread_count=-1
)

model.fit(X_train, y_train, eval_set=(X_train, y_train), verbose=100, early_stopping_rounds=100, plot=False)
```

```python
model.get_feature_importance(prettified=True)
```

```python
feature_importances = model.feature_importances_
sorteda = np.argsort(feature_importances)
fig = plt.figure(figsize=(12, 8))
plt.barh(range(len(sorteda)), feature_importances[sorteda], align='center')
plt.yticks(range(len(sorteda)), np.array(X.columns)[sorteda])
plt.title('Feature_importance')
plt.tight_layout();
```

```python
from sklearn.inspection import permutation_importance
```

```python
perm_importance = permutation_importance(model, X_test, y_test, n_repeats=10, random_state=42)
sorting = perm_importance.importances_mean.argsort()
fig = plt.figure(figsize=(12, 8))
plt.barh(range(len(sorting)), perm_importance.importances_mean[sorting], align='center')
plt.yticks(range(len(sorting)), np.array(X.columns)[sorting])
plt.tight_layout();
```

```python
from boruta import BorutaPy
from sklearn.ensemble import RandomForestClassifier
```

```python
forest = RandomForestClassifier(
   n_jobs = -1, 
   max_depth = 3
)
boruta = BorutaPy(
   estimator = forest, 
   n_estimators = 'auto',
   max_iter = 5, # number of trials to perform
    verbose=1
)
### fit Boruta (it accepts np.array, not pd.DataFrame)
boruta.fit(np.array(X_train), np.array(y_train))
### print results
green_area = X_train.columns[boruta.support_].to_list()
blue_area = X_train.columns[boruta.support_weak_].to_list()
red_area = X_train.columns[~(boruta.support_ | boruta.support_weak_)].to_list()
print('features in the green area:', green_area)
print('features in the blue area:', blue_area)
print('features in the red area:', red_area)
```

```python
target_columns = ['target_reg', 'target_class']
```

```python
X = df.drop(target_columns, axis=1, errors = 'ignore')
y = df['target_reg']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

```python
from catboost import Pool, CatBoostClassifier, CatBoostRegressor
```

```python
model = CatBoostRegressor(random_state=42, thread_count=-1)

model.fit(X_train, y_train, eval_set=(X_test, y_test), verbose=False, early_stopping_rounds=100, plot=False)
```

```python
from sklearn.inspection import permutation_importance
```

```python
perm_importance = permutation_importance(model, X_test, y_test, n_repeats=10, random_state=42)

sorting_ind = perm_importance.importances_mean.argsort()
plt.barh(range(len(sorting_ind)), perm_importance.importances_mean[sorting_ind], align='center')
plt.yticks(range(len(sorting_ind)), np.array(X.columns)[sorting_ind])
plt.tight_layout();
```
