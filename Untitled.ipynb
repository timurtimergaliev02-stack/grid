```python
from langchain_core.runnables import RunnableLambda, RunnableSequence
```

```python
runnable1 = RunnableLambda(lambda x: x+3)
runnable2 = RunnableLambda(lambda x: x+2)

Runnable = RunnableSequence(runnable1, runnable2)
print(Runnable.invoke(2))
```

```python
runnable1 = RunnableLambda(lambda x: x+3)
runnable2 = RunnableLambda(lambda x: x+2)

output1 = runnable1.invoke(2)
output2 = runnable2.invoke(output1)
print(output2)
```

```python
# from langchain_core.runnables import RunnableLambda
runna1 = RunnableLambda(lambda x: x*2)
runna2 = RunnableLambda(lambda x: x*3)

chain = runna1 | runna2

print(chain.invoke(2))
```

```python
runn1 = RunnableLambda(lambda x: x*x)
runn2 = RunnableLambda(lambda x: x*x)

chains = runn1 | runn2

print(chains.batch([1, 2, 3]))
```

```python
runn1 = RunnableLambda(lambda x: x+1)
runn2 = RunnableLambda(lambda x: x*x)

chainges = RunnableLambda(lambda x: runn1 if x % 2 else runn2)

print(chainges.batch([1, 2, 3]))
```

```python
def runnable_x(x):
    for y in x:
        yield str(y)*2

from langchain_core.runnables import RunnableLambda, RunnablePassthrough
runables = RunnableLambda(runnable_x)
for chain in runables.stream(range(5)):
    print('chain', chain)
```

```python
def add_one(x):
    return x + 1
```

```python
def add_oneout(x):
    if random.random() > 0.3:
        print('Неудачный вызов')
        raise ValueError('bad value')
    return x*2
```

```python
import random
```

```python
runnables = RunnableLambda(add_one) | RunnableLambda(add_oneout).with_retry(stop_after_attempt=10)
print(runnables.invoke(2))
```

```python
def failed_function(x):
    if random.random() > 0.00001:
        print('Неудачный вызов')
        raise ValueError('bad value')
    return x*2
```

```python
def change_function(x):
    return x*2
```

```python
function = RunnableLambda(add_one) | RunnableLambda(failed_function).with_retry(stop_after_attempt=10).with_fallbacks([RunnableLambda(change_function)])
print(function.invoke(2))
```

```python
from langchain_core.runnables import RunnableLambda, RunnablePassthrough

def Revu(main, through=None):
    if through:
        return {**main, **{'string': through}}
    return main
runnable = RunnableLambda(Revu)
chain = runnable.bind(through='int')
print(chain.invoke({'bar': 'cum'}))
```

```python
def fake_llm(defined):
    return {'origin': defined, 'Answer': 'templates'}

chain = RunnableLambda(fake_llm) | {
    'orig': RunnablePassthrough(),
    'Templates': lambda text: text['Answer'][::-1]
}
print(chain.invoke('destrict'))
```
