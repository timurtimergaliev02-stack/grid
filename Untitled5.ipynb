```python
from langchain.agents import create_agent
from langchain.tools import tool
from langchain_openai import ChatOpenAI
```

```python
@tool('get_weather', description='Return weather information for a given city', return_direct=True)
def weather_information(city: str):
    response = requests.get(f'https://wttr.in/{city}?format=j1')
    return response.json()
```

```python
llm = ChatOpenAI(
    model="gpt-4o-mini",
    base_url="https://api.vsellm.ru/" #https://vsellm.ru/
)
```

```python
model = create_agent(model=llm, tools=[weather_information], system_prompt="You are a helpful assistant",)
```

```python
Answer = model.invoke({
    'messages': [{'role': 'user', 'content': "Can you give weather information about Surgut?"}]
})
```

```python
print(Answer)
```

```python
print(Answer['messages'][-1].content)
```

```python
from langchain.messages import HumanMessage, AIMessage, SystemMessage
```

```python
Conversation = [
    SystemMessage('You are a helpful assistant! Your name is Bob.'),
    HumanMessage('What is Python'),
    AIMessage('Python is interpretator and high-lever programming language'),
    HumanMessage('When was it released?')
]
```

```python
memory = llm.invoke(Conversation)
```

```python
print(memory.content)
```
